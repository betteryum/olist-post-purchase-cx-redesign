你要做 3 张 risk table。

A. Seller risk table
输出：

outputs/tables/seller_risk_table.csv
字段建议：

字段	含义
primary_seller_id	seller
order_count	订单数
late_order_count	延迟订单数
late_rate	延迟率
avg_delay_days	平均延迟天数
bad_review_rate	1–2 星差评率
late_bad_review_count	既 late 又 bad review 的订单数
risk_segment	High volume high risk / High rate low volume 等
你的项目不是只 ranking seller，而是要把 seller risk 转成 governance action。文档里明确写 seller priority matrix 要用 late rate 和 late volume。

B. Category risk table

order_count

avg_delay_days

你要找什么支撑？
找这些：

1. seller 是按 late rate 排，还是 late volume 排
2. 有没有 seller threshold，比如 minimum order count
3. category 是按 order count / revenue / late rate / review score 哪个维度看
4. region 是看 customer_state，seller_state，还是 seller_state → customer_state lane
5. 有没有风险分层逻辑
你的额外动作
你不要只做 top 20 ranking。你要做 risk segmentation：

High volume + high late rate = priority governance group
High volume + low late rate = stable high-volume group
Low volume + high late rate = monitor / sample review group
Low volume + low late rate = low priority
这和 Gap Matrix 的方向一致：seller/category/region risk 是 V1 必做，而且目的是转成 intervention logic。
